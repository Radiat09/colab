# 2.5 Density Graph ( Things we need to know for KDE Plot)

### 2.5 Density Graph (KDE Plot) — Simple Notes

#### 1. What is it & How do we make it ourselves?
* **What it is:** A smooth line that shows the shape of your numbers (just like a histogram, but with a smooth curve instead of bars).
* **How to build it by hand (Lecture Example):**
  1. Group numbers into small ranges (e.g., student heights: `3–4 ft = 5`, `4–5 ft = 8`, `5–6 ft = 7`; Total = `20`)[cite: 1].
  2. Find the probability of each group (`group count ÷ total`). The total area under the curve always adds up to **1**.
  3. Draw a smooth bell curve over the points and connect them into one continuous line.
* **Code:** `sns.kdeplot(data=df, x='column_name', fill=True)`

---

#### 2. Why is it important? (What can we learn from it?)
* **See the big picture:** Quickly tells you where most values sit (the highest peak) and whether the data is balanced (bell-shaped) or tilted to one side (skewed).
* **Spot and remove outliers:** Very long, flat tails on the left or right show extreme/weird values that might need to be removed.

---

#### 3. How it helps with other ML steps (The Backbone)
Looking at this graph helps you decide what to do next with your data:

* **Z-Score (Finding Outliers):**
  * Works best when the graph looks like a clean **bell shape (Normal distribution)**.
  * Helps catch extreme values that are too far from the average (usually beyond $\pm 3$ standard deviations).

* **Imputation (Filling Missing Values):**
  * **If the graph is balanced (bell-shaped):** Fill missing values with the **Mean** (average).
  * **If the graph is tilted/skewed (has outliers):** Fill missing values with the **Median** (middle value), because extreme numbers will ruin the Mean.
  * *Quick check:* Plot the KDE graph before and after filling missing values to make sure the overall shape didn't change too much.
