```python
import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

## NumPy Q2.

```python
max_sal = emp_income.max()
min_sal = emp_income.min()
avg_sal = emp_income.mean()

avg_age = emp_ages.mean()

print(emp_income)
print(emp_ages)

print("Maximum Salary: ",max_sal)
print("Minimum Salary: ",min_sal)
print("Average Salary: ",avg_sal)
print("Avarage Age   : ",avg_age)
```

## NumPy Q3.

```python


age25_40 = emp_ages[(emp_ages >= 25) & (emp_ages <= 40) & (emp_income > 50000)]

satisfy_condition = age25_40.size

print("Filtered Values: ",age25_40)
print("Employees satisfy the condition: ",satisfy_condition)

```

## NumPy Q4.

```python
newAges = np.where(emp_ages<avg_age, emp_ages + 4, emp_ages + 2 )

print("Original array:", emp_ages)
print("\n")
print("Updated array:", newAges)
print("\n")
print("Average age used for condition:", avg_age)
```

## NumPy Q5.

```python
# print(emp_ids)
# print(emp_ages)
# print(emp_income)
# print(emp_roles)

# "Software Developer", "Backend Developer"


old_total_expense = emp_income.sum()

new_salaries = emp_income.astype(float).copy()

for i in range(len(emp_roles)):
    if emp_roles[i] == "Software Developer" or emp_roles[i] == "Backend Developer":
        new_salaries[i] = emp_income[i] * 1.08
    else:
        new_salaries[i] = emp_income[i]

new_total_expense = new_salaries.sum()

expense_increase = new_total_expense - old_total_expense

# print(new_salaries)
print("Old Total Salary Expense   -   -  :",old_total_expense)
print("New Total Salary Expense   -   -  :",new_total_expense)
print("Salary Increase Expense Difference:",expense_increase)

```

## Pandas Q10

```python
df['old_salary'] = df['emp_salary']
df['new_salary'] = df['emp_salary'].astype(float)

tech_roles = ["Software Developer", "Backend Developer"]

df.loc[df["role"].isin(tech_roles),"new_salary"] = df["emp_salary"] * 1.12

df.loc[~df["role"].isin(tech_roles),"new_salary"] = df["emp_salary"] * 1.05

df["increase_of_salary"] = df["new_salary"] - df["old_salary"]

result_df = df[['role', 'old_salary', 'new_salary', 'increase_of_salary']]
print(result_df)
```
